```python
import numpy as np
```

```python
# Columns: [Age, Math Marks, Science Marks]
data = np.array([
    [18, 85, 78],   # Student 1
    [19, 92, 88],   # Student 2
    [17, 76, 95],   # Student 3
    [18, 65, 70],   # Student 4
    [20, 90, 85]    # Student 5
])
```

```python
#Get the shape of the matrix.
print("Shape of the data :", np.shape(data))
```

```python
#Find the average age of students.
print("Average age of the student :", np.mean(data[:, 0]))
```

```python
#Extract Math marks of all students.
print("Maths Marks of all students :")
data[:, 1]
```

```python
#Find the highest Science mark.
print("Highest science marks :", np.max(data[:, 2]))
```

```python
#Get details of the student who scored more than 90 in Math.
print("Student details who score more than 90 in math :")
print("Age, Maths, Science")
data[data[:, 1] > 90]
```

```python
#Increase Math marks of all students by 5.
print("After increase 5 marks to all the student in maths :")
data[:, 1] + 5
```

```python
#Find how many students are younger than 19.
print("No. of students who are younger than 19 :", len(data[data[:, 0]<19]))
```

```python
#Calculate the average marks in each subject (column-wise mean).
print("Avg. marks of Maths :", np.mean(data[:, 1]))
print("Avg. marks of Science :", np.mean(data[:, 2]))
```

```python
#Get data of students who scored at least 80 in both subjects.
print("Student details who score atleast 80 in both subjects:")
print("Age Math Science")
data[(data[:, 1] >= 80) & (data[:, 2]>=80)]
```

```python
#Replace all Science marks < 75 with 0.
print("Replace all Science marks < 75 with 0")
data[:, 2][data[:, 2]<75] = 0
data
```
