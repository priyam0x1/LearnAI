```python
import numpy as np
import pandas as pd
```

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
df
```

```python
# Checking missing value
df.isna()
```

```python
# Missing value count in each column
df.isna().sum()
```

```python
# Is there any missing value in each column or not
df.isna().any()
```

```python
df
```

```python
#Droping null value works through row
df.dropna()
```

```python
#if atleast n non null values present then this will not remove
df.dropna(thresh=3)
```

```python
#Fill null values with 0
df.fillna(0)
```

```python
#Filling null values differently in each column
values = {'A':1, 'B':100, 'C':200, 'D':300}
df.fillna(value=values)
```

```python
df
```

```python
#Filling with Mean
df.fillna(df.mean())
```
